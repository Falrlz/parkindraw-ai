# ParkinDraw AI --- Notebook 01: Dataset Audit & EDA (NewHandPD)

Notebook ini menyajikan **Exploratory Data Analysis (EDA)** dan **Audit Integritas Data** untuk dataset NewHandPD.

```python
import json

import matplotlib.pyplot as plt
import pandas as pd

# Load metadata artifacts
df_images = pd.read_csv('../data/metadata/images.csv')
df_subjects = pd.read_csv('../data/metadata/subjects.csv')
df_duplicates = pd.read_csv('../data/metadata/duplicate_groups.csv')

with open('../data/metadata/audit_report.json') as f:
    report = json.load(f)

print(f"Total Citra    : {len(df_images)}")
print(f"Total Subjek   : {len(df_subjects)}")
print(f"Healthy (0)    : {report['healthy_subjects']}")
print(f"Parkinson (1)  : {report['parkinson_subjects']}")
```

## 1. Distribusi Kelas & Subjek

Menampilkan distribusi subjek Healthy vs Parkinson.

```python
df_subjects['class_name'].value_counts().plot(kind='bar', color=['#2ecc71', '#e74c3c'])
plt.title('Distribusi Subjek NewHandPD (Healthy vs Parkinson)')
plt.ylabel('Jumlah Subjek')
plt.xlabel('Kelas')
plt.show()
```
